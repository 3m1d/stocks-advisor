# TFT модель для предсказания стоимости акций

TFT (Temporal Fusion Transformer) - модель анализа временных рядов на основе архитектуры "трансформер". Эта модель может прогнозировать сразу весь необходимый нам горизонт, без накопления ошибки. А ещё мы можем обучить всего одну модель на нескольких таргетах сразу, то есть будет одна модель для нескольких акций.

План такой:
- Основные признаки + таргет: GAZP, LKOH, ROSN, SBER
- Внешние признаки: USDRUB, EURRUB, CNYRUB, GLDRUB, Brent, IMOEX
